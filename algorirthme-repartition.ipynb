L'ancien choixpeau magique classait les élèves dans les différentes maisons en fonction de certaines de leurs caractéristiques psychologiques. Le choix se faisait de la manière suivante : 
- Gryffondor : si l'élève privilégiait le courage, la bravoure et la défense des valeurs,
- Serdaigle : si l'élève valorisait l'intelligence, la sagesse et la créativité,
- Poufsouffle : si l'élève mettait en avant la loyauté, le travail acharné et la bienveillance,
- Serpentard : si l'élève accordait de l’importance à l’ambition, la ruse et la détermination.

La professeur Sibylle Trelawney propose alors un questionnaire à 4 questions seulement :
- Question 1 : À quel point aimez-vous prendre des risques pour défendre une cause juste ?
- Question 2 : Comment évaluez-vous votre curiosité intellectuelle et votre soif d’apprendre ?
- Question 3 : Quelle importance accordez-vous à la loyauté envers vos amis et votre famille ?
- Question 4 : À quel point êtes-vous ambitieux et déterminé à atteindre vos objectifs personnels ? 

Les élèves doivent alors répondre à chaque question avec un score compris entre 1 (pas du tout) et 10 (complètement) représentant sa réponse. Le choix se fait en affectant l'élève à la maison :

- Poufsouffle : si l'élève a mis le meilleur score à la Question 3,
- Serpentard : si l'élève a mis le meilleur score à la Question 4.

**Remarque :** Même si la magie ne peut pas répartir les élèves, elle peut agrandir ou rétrécir sans souci les chambres et les salles de classe donc il n'y a pas de contrainte liée au nombre d'élèves affectés à chaque maison. Il vaut mieux répartir correctement chaque élève sans affecter le même nombre d'élèves à chaque maison plutôt que mal répartir les élèves de manière équitable !

Concrètement, on décide de stocker les réponses au questionnaire dans un tableau de chaînes de caractères et d'entiers contenant successivement un élève (sous forme d'une chaîne de caractères), et les quatres réponses aux questions sous forme d'entiers. Un tel tableau est appelé *tableau de réponses*. On suppose qu'un tableau de réponses est toujours correctement construit (une chaîne de caractères pour chaque élève et 4 entiers entre 1 et 10 pour les réponses).

### Question 3

Définissez la fonction `lecture_reponses` prenant en paramètre un nom de fichier et retournant un tableau de réponses en fonction des données contenues dans le fichier. Ce dernier comporte une ligne par élève. Chaque ligne a le format `nom:R1/R2/R3/R4` où `nom` correspond au nom de l'élève et `R1`, `R2`, `R3`, `R4` correspondent aux scores (réponses aux questions) de l'élève. Par exemple, l'appel de la fonction `lecture_reponses` avec un nom de fichier contenant 
```
Harry Potter:10/5/8/1
Cedric Diggory:6/7/9/4
Drago Malefoy:1/3/2/10
```
doit retourner le tableau de réponses donné en exemple.

Définissez une fonction de tests unitaires pour la fonction `lecture_reponses`.

````python
def lecture_reponses(fich):   # Initialisation de la fonction "lecture_reponses"
    
    tab = [] # Initalisation du tableau "tab"
    f = open(fich, "r")  # Ouvre le fichier en argument et le lit stocké dans une variable "f"
    lignes = f.readlines()  # Crée un tableau "lignes" avec le contenu de chaque ligne du fichier
    f.close()  # Ferme le fichier

    i = 0     # Initialisation de la variable i 
    while i < len(lignes): # Création de la boucle "tant que i est strictement inférieur à la longueur du tableau "lignes"
        
        ligne = lignes[i].strip()  # Enlève les espaces et retour à la ligne à chaque élement d'indice "i" du tableau
        
        if ligne != "":  # Condition qui valide si la ligne n'est pas vide
            
            nom_et_notes = ligne.split(":") # Initialisation du tableau "nom_et_notes" qui découpe la ligne à chaque ":"
            nom = nom_et_notes[0]  # Initialisation du tableau "nom" qui prend le premier élément du tableau "nom_et_notes"
            notes = nom_et_notes[1].split("/") # Initialisation du tableau "notes" qui découpe le 2ème élément du tableau nom_et_notes à chaque "/"
            tab.append(nom)  # Ajout de la variable "nom" dans le tableau "tab"

            j = 0    # Initialisation de la variable "j"
            while j < len(notes): # Création de la boucle "tant que j est strictement inférieur à la longueur du tableau "notes"
                
                tab.append(int(notes[j]))  # Ajout de l'élément à l'indice "j" du tableau "notes" dans le tableau "tab
                j += 1 # Incrémentation de 1 à la variable "j"

        i += 1 # Incrémentation de 1 à la variable "i"

    return tab # Retour du tableau "tab"

def test_lecture_reponses(): # Initialisation de la fonction "test_lectures_reponses()
    resultat = lecture_reponses("questionnaire_premiere_annee.txt")  # Initialisation de la variable "resultat" qui appelle la fonction
    assert (lecture_reponses("questionnaire_premiere_annee.txt") == resultat) == True # Test 1 : Cas ou l'appel de la fonction est indentique à la variable résultat
    assert (lecture_reponses("questionnaire_premiere_annee.txt") != resultat) == False # Test 2 : Cas ou il y'a une erreur
    print("Test de la fonction lecture_reponses : ok") # Affiche "Test de la fonction lecture_reponses : ok" si les tests sont réussis

test_lecture_reponses() # Appel de la fonction "test_lecture_reponses
lecture_reponses("questionnaire_premiere_annee.txt") # Appel de la fonction "lecture_reponses" avec comme paramètre le nom de fichier questionnaire_premiere_annee.txt"
````

### Question 4

Définissez la fonction `maison` prenant en paramètre un tableau de réponses ainsi que l'indice d'un élève. La fonction retourne la maison correspondant à l'affectation de cet élève. Ainsi, l'appel de la fonction maison avec le tableau de réponses donné en exemple doit retourner : 
- `"Gryffondor"` avec l'indice `0` (correspondant à Harry Potter),
- `"Poufsouffle"` avec l'indice `5` (correspondant à Cedric Diggory),
- `"Serpentard"` avec l'indice `10` (correspondant à Drago Malefoy).

En effet, Harry Potter a mis le meilleur score à la question 1 et doit donc être affecté à Gryffondor. Cedric Diggory a mis le meilleur score à la question 3 et doit être affecté à la maison Poufsouffle. Finalement, Drago a mis le meilleur score à la question 4 et doit donc être affecté à la maison Serpentard.

Définissez également une fonction de tests unitaires pour la fonction `maison`.

````python
def maison_eleve(tab, indice):  # Initialisation de la fonction "maison_eleve" prenant en compte 2 arguments
    
    maisons = ["Gryffondor", "Serdaigle", "Poufsouffle", "Serpentard"] # Initialisation du tableau "maisons"
    index = indice + 1 # Initialisation de la variable "index" étant le deuxième paramètre auquel on incrémente de 1
    valeur_max = tab[index] # Initialisation de la variable "valeur_max" étant le tableau "tab" à l'élément de l'indice du premier paramètre
    index_max = 0 # Initialisation de la variable "indice_max" 
    
    while index < indice + 5: # Création de la boucle "tant que index est strictement inférieur à indice + 5
 
        if tab[index] >= valeur_max: # Condition valide si tab à l'élément de l'indice "index" est supérieur ou égal à la variable "valeur_max"
            valeur_max = tab[index]  # Ajout de l'élement à l'indice "index" du tableau "tab" dans la variable "valeur_max"
            index_max = index   # Ajout de la variable "index" à la variable "index_max"
        index += 1 # Incrémentation de 1 à la variable "index"
            
    return maisons[index_max - indice - 1] # Retour du tableau "maisons" à l'élement de l'indice "index_max - indice - 1"
        


def test_maison_eleve(): # Initialisation de la fonction "test_maison_eleve"
    assert (maison_eleve(reponses,10)=="Serpentard") == True # Test 1 : Cas ou l'élève est affecté à Serpentard
    assert (maison_eleve(reponses,5)=="Poufsouffle") == True # Test 2 : Cas ou l'élève est affecté à Poufsouffle
    assert (maison_eleve(reponses,5)=="Gryffondor") == False # Test 3 : Cas ou il y'a une erreur
    print("Test de la fonction maison_eleve : ok") # Affiche "Test de la fonction lecture_reponses : ok" si les tests sont réussis


test_maison_eleve() # Appel de la fonction "test_maison_eleve"
maison_eleve(reponses,10) # Appel de la fonction "maison_eleve" avec comme paramètres le tableau "reponses" et "10"
````

````python
from random import * # Utilisation du module "random"

from json import * # Utilisation du module "json"

reponse = lecture_reponses("questionnaire_premiere_annee.txt") # Initialisation de la variable "reponse" correspondant à la fonction "lecture_reponses" ayant comme paramètre "questionnaire_premiere_annee.txt"
eleve = eleves(reponse) # Initialisation de la variable "eleve" correspondant à la fonction "eleves" avec comme paramètre la variable "reponse"

f = open("affectation_premiere_annee.json", "r")  # Lecture du fichier "affectation_premiere_annee.json" stocké dans la variable "f"
dico1 = json.load(f) # Création d'un dictionnaire "dico1" dont le contenu est celui de la variable "f"
f.close()  # Fermeture du fichier 

def maison_aleatoire(): # Initialisation de la fonction maison_aleatoire():
    i = 0 # Initialisation de la variable "i"
    erreurs = 0 # Initialisation de la variable "erreurs"
    while i < 100: # Création de la boucle "tant que i est strictement inférieur à 100"
        
        for element in dico1: # Création d'une boucle avec la variable "element" qui se répète pour chaque élément de dico1
            dico = {} # Initialisation du dictionnaire "dico"
            alea = randint(1,4) # Initialisation de la variable "alea" correspondant à un un nombre aléatoire entre 1 et 4
            
            if alea == 1: # Condition valide si la variable aléa est égale à 1
                dico[eleve[i]]="Gryffondor" # Ajout au dictionnaire "dico" la clé correspondant à l'élément du tableau "eleve" à l'indice "i" et ayant comme valeur "Gryffondor"
            
            elif alea == 2: # Condition valide si la variable aléa est égale à 2
                dico[eleve[i]]="Poufsouffle" # Ajout au dictionnaire "dico" la clé correspondant à l'élément du tableau "eleve" à l'indice "i" et ayant comme valeur "Poufsouffle"
            
            elif alea == 3: # Condition valide si la variable aléa est égale à 3
                dico[eleve[i]]="Serdaigle" # Ajout au dictionnaire "dico" la clé correspondant à l'élément du tableau "eleve" à l'indice "i" et ayant comme valeur "Serdaigle"
            
            else: # Condition valide si la variable aléa est égale à 4
                dico[eleve[i]]="Serpentard" # Ajout au dictionnaire "dico" la clé correspondant à l'élément du tableau "eleve" à l'indice "i" et ayant comme valeur "Serpentard"
            
            erreurs += nb_erreurs(dico, dico1) # Incrémentation de la valeur retournée par la fonction "nb_erreurs" avec comme paramètres "dico" et "dico1" à la variable "erreurs"
            
        i += 1 # Incrémentation de 1 à la variable "i"
    return erreurs # Retour de la variable "erreurs"

moyenne_erreurs = maison_aleatoire() / 100 # Divise la valeur retournée par la fonction "maison_aléatoire" par 100
pourcentage_erreurs = (moyenne_erreurs / len(dico1)) * 100 # Divise la variable "moyenne_erreurs" par la longueur du dictionnaire "dico1" puis multipliée par 100 
print("Le pourcentage d'erreur est de ", pourcentage_erreurs, "%") # Affiche la variable "pourcentage d'erreur"


# On en conclut que le pourcentage d'erreur avec cette méthode de répartition est plus élevée que la méthode précédente puisque cette méthode à un pourcentage d'erreur d'environ 75% tandis que la méthode précédente à un pourcentage d'erreur d'environ 57%;
# On peut donc dire que la méthode précédente est la plus efficace;
````
